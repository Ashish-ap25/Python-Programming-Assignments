```python
# 9. Palindromic Substrings with Pattern Listing
#
# You are given a string s consisting of lowercase English letters.
# A palindrome is a string that reads the same forward and backward.
# A substring is a contiguous sequence of characters within the string.
#
# Your Task:
# 1. Count the total number of palindromic substrings in the string.
# 2. Also print all palindromic substrings (patterns).
# Duplicate substrings at different positions should be counted separately.
#
# Constraints:
# • 1 ≤ s.length ≤ 1000
# • s consists only of lowercase English letters
#
# Example 1:
# Input: s = "ababa"
# Output:
# Total Count: 9
# Palindromic Substrings:
# a
# b
# a
# b
# a
# aba
# bab
# aba
# ababa
#
# Explanation:
# Single character palindromes = 5
# Length 3 palindromes = 3
# Length 5 palindrome = 1
# Total = 5 + 3 + 1 = 9
#
# Example 2:
# Input: s = "racecar"
# Output:
# Total Count: 10
# Palindromic Substrings:
# r
# a
# c
# e
# c
# a
# r
# cec
# aceca
# racecar
#
# Explanation:
# Single characters = 7
# "cec" = 1
# "aceca" = 1
# "racecar" = 1
# Total = 10
#
# Example 3:
# Input: s = "level"
# Output:
# Total Count: 7
# Palindromic Substrings:
# l
# e
# v
# e
# l
# eve
# level
#
# Explanation:
# Single characters = 5
# "eve" = 1
# "level" = 1
# Total = 7
```

```python
#Find and count all substrings that read the same forward and backward. Print every palindrome, including duplicates occurring at different positions.
```

```python
s = input("Enter a string: ")

palindromes = []

for i in range(len(s)):
    for j in range(i + 1, len(s) + 1):

        sub = s[i:j]

        if sub == sub[::-1]:
            palindromes.append(sub)

print("Total Count:", len(palindromes))

print("Palindromic Substrings:")

for p in palindromes:
    print(p)
```
