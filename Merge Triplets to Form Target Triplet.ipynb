```python
# 8. Merge Triplets to Form Target Triplet
#
# A triplet is an array of three integers. You are given a 2D integer array triplets,
# where triplets[i] = [ai, bi, ci] describes the ith triplet.
#
# You are also given an integer array target = [x, y, z] that describes the triplet
# you want to obtain.
#
# To obtain target, you may apply the following operation on triplets any number
# of times (possibly zero):
#
# • Choose two indices (0-indexed) i and j (i != j) and update triplets[j] to become
#   [max(ai, aj), max(bi, bj), max(ci, cj)].
#
# • For example, if triplets[i] = [2, 5, 3] and triplets[j] = [1, 7, 5],
#   triplets[j] will be updated to [max(2, 1), max(5, 7), max(3, 5)] = [2, 7, 5].
#
# Return true if it is possible to obtain the target triplet [x, y, z] as an element
# of triplets, or false otherwise.
#
# Constraints:
# • 1 <= triplets.length <= 10^5
# • triplets[i].length == target.length == 3
# • 1 <= ai, bi, ci, x, y, z <= 1000
#
# Example 1:
# Input: triplets = [[2,5,3],[1,8,4],[1,7,5]], target = [2,7,5]
# Output: true
#
# The merging triplet to reach the target = [[2,5,3],[1,8,4]]
# Target = [2,7,5]
#
# Explanation:
# Perform the following operations:
# - Choose the first and last triplets [[2,5,3],[1,8,4],[1,7,5]].
#   Update the last triplet to be [max(2,1), max(5,7), max(3,5)] = [2,7,5].
#   triplets = [[2,5,3],[1,8,4],[2,7,5]]
# The target triplet [2,7,5] is now an element of triplets.
#
# Example 2:
# Input: triplets = [[3,4,5],[4,5,6]], target = [3,2,5]
# Output: false
#
# The merging triplet to reach the target = [[3,4,5],[4,5,6]]
# Target = none
#
# Explanation:
# It is impossible to have [3,2,5] as an element because there is no 2
# in any of the triplets.
#
# Example 3:
# Input: triplets = [[2,5,3],[2,3,4],[1,2,5],[5,2,3]], target = [5,5,5]
# Output: true
#
# The merging triplet to reach the target = [[2,5,5],[5,2,3]]
# Target = [5,5,5]
#
# Explanation:
# Perform the following operations:
# - Choose the first and third triplets.
#   Update the third triplet to be [max(2,1), max(5,2), max(3,5)] = [2,5,5].
# - Choose the third and fourth triplets.
#   Update the fourth triplet to be [max(2,5), max(5,2), max(5,3)] = [5,5,5].
# The target triplet [5,5,5] is now an element of triplets.
```

```python
#Check whether the given triplets can be combined using maximum values to form the exact target triplet.
```

```python
import ast

triplets = ast.literal_eval(input("Enter triplets: "))
target = ast.literal_eval(input("Enter target: "))

found = [False, False, False]

for a, b, c in triplets:

    if a <= target[0] and b <= target[1] and c <= target[2]:

        if a == target[0]:
            found[0] = True

        if b == target[1]:
            found[1] = True

        if c == target[2]:
            found[2] = True

print(str(all(found)).lower())
```
