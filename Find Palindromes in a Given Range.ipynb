```python
# 2. Find Palindromes in a Given Range
# • You are given two integers m and n representing a range.
# • Your task is to find and return all numbers within the range [m, n] that are palindromes.
# • A number is a palindrome if it reads the same forward and backward.
# Constraints:
# • 0 ≤ m ≤ n ≤ 10⁶
# Example 1:
# Input: m = 10, n = 50
# Output: [11, 22, 33, 44]
# Explanation: These numbers remain the same when reversed.
# Example 2:
# Input: m = 100, n = 150
# Output: [101, 111, 121, 131, 141]
# Explanation: Each number reads the same from both directions.
# Example 3:
# Input: m = 1, n = 9
# Output: [1,2,3,4,5,6,7,8,9]
# Explanation: All single-digit numbers are palindromes.
```

```python
#Find all numbers within a given range that remain the same when their digits are reversed.
```

```python
m = int(input("Enter starting number: "))
n = int(input("Enter ending number: "))

palindromes = []

for num in range(m, n + 1):
    if str(num) == str(num)[::-1]:
        palindromes.append(num)

print("Palindrome numbers:", palindromes)
```
