```python
# 1. Single Number: Given a non-empty array of integers nums, every element
# appears twice except for one. Find that single one. You must implement a solution with a
# linear runtime complexity and use only constant extra space.
# Constraints:
# • 1 <= nums.length <= 3 * 104
# • -3 * 104 <= nums[i] <= 3 * 104
# • Each element in the array appears twice except for one element which appears only
# once.
# Example 1:
# Input: nums = [2,2,1]
# Output: 1
# Example 2:
# Input: nums = [4,1,2,1,2]
# Output: 4
# Example 3:
# Input: nums = [1]
# Output: 1
```

```python
#Find the number that appears only once in an array where every other number appears exactly twice.
```

```python
nums = list(map(int, input("Enter the array elements: ").split()))

single = 0

for num in nums:
    single = single ^ num

print("Single Number:", single)
```
