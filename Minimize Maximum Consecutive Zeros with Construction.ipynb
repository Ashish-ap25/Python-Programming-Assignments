```python
# 7. Minimize Maximum Consecutive Zeros with Construction
#
# You are given two integers:
# • L → length of a binary string
# • K → number of ones (1s) in the string
#
# The remaining positions (L − K) are filled with zeros (0s).
#
# Your tasks are:
# 1. Generate all possible binary strings of length L containing exactly K ones.
# 2. For each string, determine the length of the longest contiguous block of zeros.
# 3. Identify the optimal arrangement(s) such that this maximum zero block is minimized.
#
# Output Requirements:
# • Print all possible valid binary strings
# • For each string, print its maximum consecutive zero length
# • Print the optimal string(s) that achieve the minimum possible longest zero block
# • Finally, print that minimum length
#
# Constraints:
# • 1 ≤ L ≤ 15 (limited due to combinatorial explosion)
# • 0 ≤ K ≤ L
#
# Input Format:
# A single line containing two integers: L K
#
# Output Format:
# • List all possible binary strings
# • For each string:
#   → string → max zero block
# • Then print:
#   → Optimal string(s)
#   → Minimum longest zero block
#
# Example 1:
# Input: 4 2
# Output:
# Possible Strings:
# 1100 → max zeros = 2
# 1010 → max zeros = 1
# 1001 → max zeros = 2
# 0110 → max zeros = 1
# 0101 → max zeros = 1
# 0011 → max zeros = 2
#
# Optimal Arrangement(s):
# 1010, 0110, 0101
# Minimum Longest Zero Block: 1
#
# Explanation:
# Strings where zeros are evenly distributed have the smallest maximum zero block.
#
# Example 2:
# Input: 5 0
# Output:
# Possible Strings:
# 00000 → max zeros = 5
# Optimal Arrangement(s):
# 00000
# Minimum Longest Zero Block: 5
#
# Explanation:
# All positions are zeros, so the longest block is the full length.
#
# Example 3:
# Input: 5 5
# Output:
# Possible Strings:
# 11111 → max zeros = 0
# Optimal Arrangement(s):
# 11111
# Minimum Longest Zero Block: 0
#
# Explanation:
# No zeros present.
```

```python
#Generate all binary strings containing exactly K ones and find the arrangement where the longest consecutive sequence of zeros is as small as possible.
```

```python
from itertools import combinations

L, K = map(int, input().split())

strings = []

for positions in combinations(range(L), K):
    s = ['0'] * L

    for i in positions:
        s[i] = '1'

    strings.append(''.join(s))


def max_zeros(s):
    count = 0
    maximum = 0

    for ch in s:
        if ch == '0':
            count += 1
            maximum = max(maximum, count)
        else:
            count = 0

    return maximum


print("Possible Strings:")

results = []

for s in strings:
    zeros = max_zeros(s)
    results.append((s, zeros))
    print(s, "→ max zeros =", zeros)


minimum = min(zeros for s, zeros in results)

optimal = [s for s, zeros in results if zeros == minimum]

print("Optimal Arrangement(s):")
print(", ".join(optimal))

print("Minimum Longest Zero Block:", minimum)
```
