```python
# 5. Balanced Binary Tree
#
# Given a binary tree, determine if it is height-balanced.
#
# Constraints:
# • The number of nodes in the tree is in the range [0, 5000].
# • -10^4 <= Node.val <= 10^4
#
# Example 1:
# Input: root = [3, 9, 20, null, null, 15, 7]
# Output: true
#
# Example 2:
# Input: root = [1, 2, 2, 3, 3, null, null, 4, 4]
# Output: false
#
# Example 3:
# Input: root = [4]
# Output: true
```

```python
#Check whether a binary tree is height-balanced, meaning the height difference between the left and right subtrees of every node must not exceed 1
```

```python
import ast

root = input("Enter tree: ")
root = root.replace("null", "None")

root = ast.literal_eval(root)

def height(i):
    if i >= len(root) or root[i] is None:
        return 0

    left = height(2 * i + 1)
    right = height(2 * i + 2)

    if left == -1 or right == -1 or abs(left - right) > 1:
        return -1

    return 1 + max(left, right)

print(str(height(0) != -1).lower())
```
