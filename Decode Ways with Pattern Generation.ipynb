```python
# 4. Decode Ways with Pattern Generation
#
# You have intercepted a secret message encoded as a string s containing only digits.
# The message is decoded using the mapping:
# • "1" → 'A', "2" → 'B', … "26" → 'Z'
#
# A message can be decoded in multiple ways because:
# • A single digit can map to a letter
# • Two digits together can also form a valid letter (if between 10 and 26)
#
# Your Task:
# 1. Compute the total number of ways to decode the string
# 2. Also print all possible decoded strings (patterns)
#
# A decoding is valid only if:
# • No segment starts with '0'
# • Valid numbers are from 1 to 26 only
#
# Page 2 of 51:
# Constraints:
# • 1 ≤ s.length ≤ 100
# • s contains only digits
# • Output fits in 32-bit integer
#
# Example 1:
# Input: s = "12"
# Output:
# Total Ways: 2
# Decoded Strings:
# AB
# L
#
# Explanation:
# Possible groupings:
# • "1" + "2" → A B → AB
# • "12" → L
# Total = 2
#
# Example 2:
# Input: s = "226"
# Output:
# Total Ways: 3
# Decoded Strings:
# BBF
# BZ
# VF
#
# Explanation:
# • "2" + "2" + "6" → B B F → BBF
# • "2" + "26" → B Z → BZ
# • "22" + "6" → V F → VF
# Total = 3
#
# Example 3:
# Input: s = "11106"
# Output:
# Total Ways: 2
# Decoded Strings:
# AAJF
# KJF
#
# Explanation:
# Valid decodings:
# • (1,1,10,6) → A A J F → AAJF
# • (11,10,6) → K J F → KJF
# Invalid:
# • (1,11,06) → "06" not valid
#
# Example 4 (Invalid Case):
# Input: s = "06"
# Output:
# Total Ways: 0
# Decoded Strings: None
#
# Explanation:
# • "06" is invalid because leading zero is not allowed.
```

```python
#Find the total number of ways to decode a numeric string into letters (A–Z) and print all possible decoded strings.
```

```python
s = input("Enter the numeric string: ")

decoded = []

def generate(index, current):

    if index == len(s):
        decoded.append(current)
        return

    if s[index] == '0':
        return

    num = int(s[index])
    letter = chr(64 + num)

    generate(index + 1, current + letter)

    if index + 1 < len(s):
        num = int(s[index:index + 2])

        if 10 <= num <= 26:
            letter = chr(64 + num)
            generate(index + 2, current + letter)

generate(0, "")

print("Total Ways:", len(decoded))

print("Decoded Strings:")

if len(decoded) == 0:
    print("None")
else:
    for result in decoded:
        print(result)
```
