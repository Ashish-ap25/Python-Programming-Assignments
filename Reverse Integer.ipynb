```python
# 6. Reverse Integer:
# Given a signed 32-bit integer x, return x with its digits reversed.
# If reversing x causes the value to go outside the signed 32-bit
# integer range [-2^31, 2^31 - 1], then return 0.
# Assume the environment does not allow you to store 64-bit integers
# (signed or unsigned).
#
# Constraints:
# • -2^31 <= x <= 2^31 - 1
#
# Example 1:
# Input: x = 123
# Output: 321
#
# Example 2:
# Input: x = -123
# Output: -321
#
# Example 3:
# Input: x = 120
# Output: 21
```

```python
#Reverse the digits of a given integer while preserving its sign. If the reversed number exceeds the 32-bit signed integer range, return 0
```

```python
x = int(input("Enter an integer: "))

sign = -1 if x < 0 else 1

num = abs(x)
rev = int(str(num)[::-1]) * sign

if -2**31 <= rev <= 2**31 - 1:
    print(rev)
else:
    print(0)
```
